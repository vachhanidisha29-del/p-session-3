```python
import pandas as pd

sales_df = pd.read_csv("kirana_sales.csv")

first50 = sales_df.loc[0:49, ["bill_id", "product", "qty"]]

first50
```

```python
last20 = sales_df.iloc[-20:, 2:5]

last20
```

```python
big_bills = sales_df[
    (
        (sales_df["payment_mode"] == "UPI") &
        (sales_df["qty"] >= 5) &
        (sales_df["unit_price_inr"] > 200)
    )
    |
    (
        (sales_df["payment_mode"] == "Card") &
        (sales_df["qty"] >= 8)
    )
]

big_bills
```

```python
min_qty_upi = 5
min_price_upi = 200
min_qty_card = 8

big_query = sales_df.query(
    '(payment_mode == "UPI" and qty >= @min_qty_upi and unit_price_inr > @min_price_upi) or '
    '(payment_mode == "Card" and qty >= @min_qty_card)'
)

big_query
```
